**Read then write**

```python
with open("s.txt",'r+') as fp:
    d = fp.read()
    print(d)
    fp.write("\nnew content by r+ mode!!!!")
```

**Write then Read**

```python
with open("s.txt",'w+') as fp:
    fp.write("content by w+ mode!!!!")
    fp.seek(0)
    data = fp.read()
    print(data)
```

**Append (+)**

```python
with open('s.txt','a+') as fp:
    fp.write("\n content made by a+ mode!!!!")
    fp.seek(0)
    j = fp.read()
    print(j)
```
